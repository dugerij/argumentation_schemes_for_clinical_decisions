# Pipeline Check

This notebook verifies that indexing, retrieval, and the argumentation recommendation path work together from one scenario.

```python
from __future__ import annotations

import asyncio
import json
import sys
from dataclasses import asdict
from pathlib import Path

from dotenv import load_dotenv

repo_root = Path.cwd()
if not (repo_root / "main.py").exists():
    for parent in Path.cwd().resolve().parents:
        if (parent / "main.py").exists():
            repo_root = parent
            break

sys.path.insert(0, str(repo_root))
load_dotenv(repo_root / ".env")

from eval.pipeline_check import run_pipeline_check
```

```python
SCENARIO = "A patient with chronic kidney disease and hypertension needs blood pressure management."
CLINICAL_GOAL = "reduce blood pressure while avoiding renal harm"
USE_UMLS = False
SCHEMA_GUIDED = False
DRY_RUN = True
INPUT_DIR = repo_root / "data" / "evidence" / "mimic_discharge_subset"
OUTPUT_DIR = repo_root / "output" / "pipeline_check"

print("scenario:", SCENARIO)
print("clinical_goal:", CLINICAL_GOAL)
print("output_dir:", OUTPUT_DIR)
```

```python
result = asyncio.run(
    run_pipeline_check(
        input_dir=INPUT_DIR,
        output_dir=OUTPUT_DIR,
        scenario=SCENARIO,
        clinical_goal=CLINICAL_GOAL,
        use_umls=USE_UMLS,
        schema_guided=SCHEMA_GUIDED,
        dry_run=DRY_RUN,
    )
)

print(json.dumps(asdict(result), indent=2))
```
